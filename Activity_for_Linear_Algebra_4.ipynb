```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets
```

```python
x, y = datasets.make_blobs(
    n_samples=200,
    n_features=2,
    centers=[[2, 2], [10, 10]],
    center_box=(0, 10),
    cluster_std=3
)

# Convert labels from {0,1} to {-1,+1}
y[y == 0] = -1
```

```python
def show(w, w0, x, y):
    """Plot the decision boundary and data points."""
    x_val = np.linspace(-5, 10)
    # w[0] = w1, w[1] = w2
    y_val = (-w[0] * x_val - w0) / w[1]
    plt.plot(x_val, y_val)
    plt.scatter(x[:, 0], x[:, 1], c=y)
    plt.xlim(x[:, 0].min() + 0.2, x[:, 0].max() + 0.2)
    plt.ylim(x[:, 1].min() + 0.2, x[:, 1].max() + 0.2)
    plt.show()

# Random initialisation
w  = np.random.normal(size=(2,))
w0 = np.random.normal()
print(f"Initial weights: w = {w}, w0 = {w0}")

show(w, w0, x, y)
```

```python
def dist_from_hyperplane(w, w0, x):
    """
    Signed distance of point x from the hyperplane w·x + w0 = 0.
    np.linalg.norm(w) computes ||w|| = sqrt(w1^2 + w2^2).
    """
    return (np.dot(w, x) + w0) / np.linalg.norm(w)
```

```python
def perceptron_training(x, y, w, w0, num_iter=100):
    """
    Train the Perceptron.
    Update rule: w_new = w_old + xi * yi  (for misclassified xi)
                 w0_new = w0_old + sign(yi)
    Note: updating w0 only requires the sign of yi (direction, not magnitude)
          because changing w0 shifts the boundary up/down without rotation.
    """
    for i in range(num_iter):
        print(f"Iteration {i + 1}")
        misclassified = 0

        for (xi, yi) in zip(x, y):
            d  = dist_from_hyperplane(w, w0, xi)
            hs = np.sign(d)           # assigned label

            if hs != np.sign(yi):     # misclassified?
                misclassified += 1
                w  = w  + xi * yi     # Perceptron update rule
                w0 = w0 + np.sign(yi)

        accuracy = 100 - (100 * misclassified / len(y))
        print(f"  Accuracy : {accuracy:.1f}%")
        print(f"  Weights  : w1={w[0]:.4f}, w2={w[1]:.4f}, w0={w0:.4f}")
        show(w, w0, x, y) # This will show a plot for each iteration

        if misclassified == 0:
            print("✅ All points correctly classified — stopping early.")
            return w, w0

    return w, w0

w, w0 = perceptron_training(x, y, w, w0)
```
