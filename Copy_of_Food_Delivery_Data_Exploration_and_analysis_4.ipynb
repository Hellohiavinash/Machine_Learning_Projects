## <span style="color: skyblue;"> Parameters for Splitting Arrays </span>

Splitting is an important concept, especially in machine learning workflows.

### Real world intuition

When building a machine learning model, we should not train and test on the same data. Instead:

* Split data into Training data and Testing data
* Train the model on partial data
* Keep some data unseen
* Test the model on unseen data — this helps validate model performance

<span style="color: red;"> Important:</span>
* Without splitting, we cannot properly judge whether a model is good or not

<span style="color: orange;">Doubt by Learner:</span>
* Why can we not use full data for both training and testing?
* What happens if we do not split the dataset?

All splitting functions take the following parameters:

1. array: The array to be split.
2. indices_or_sections:
   - If an integer `N`, the array is split into `N` equal sections.
   - If a list of indices, the splits happen at those positions.

## <span style="color: skyblue;"> Splitting Arrays with np.split() </span>

`np.split()` splits an array into sections along a specified axis:
- **axis=0**: Split rows.
- **axis=1**: Split columns (for 2D arrays).


**Example 1: Splitting a 1D Array**

```python
import numpy as np

# 1D array
array_1d = np.array([1, 2, 3, 4, 5, 6])

# Split into 3 sections
split_1d = np.split(array_1d, 3)

print("Original Array:", array_1d)
print("Splits:", split_1d)
```

**Prerequisites for Splitting**

- The total number of elements **must be divisible** by the number of splits.
- For 2D arrays, ensure the axis matches the shape of the array.

**Example**:
- A 1D array with 6 elements can be split into 2 or 3 equal parts, but **not 4**.
- A 2D array with 4 columns can be split horizontally into 2 parts, but **not 3**.

```python
# This is expected to fail!
split_1d = np.split(array_1d, 4)
```

**Example 2: Splitting a 2D Array**

```python
# 2D array
array_2d = np.array([[1, 2, 3, 4],
                     [5, 6, 7, 8]])

# Split into 2 along axis=1 (columns)
split_2d = np.split(array_2d, 2, axis=1)

print("Original Array:\n", array_2d)
print("=="*10)
print("Splits:")
for part in split_2d:
    print(part)
```

**Explanation:**
- **1D Array**: Split into 3 equal parts.
- **2D Array**: Split columns into 2 equal sections using `axis=1`.

## <span style="color: skyblue;"> Horizontal Splitting with np.hsplit() </span>

`np.hsplit()` splits a 2D array horizontally into sections (columns).

>**Shortcut:** Equivalent to np.split(array, N, axis=1).

**Example: Horizontal Splitting**

```python
# Horizontal split
hsplit_2d = np.hsplit(array_2d, 2)

print("Original Array:\n", array_2d)
print("=="*10)
print("Horizontal Splits:")
for part in hsplit_2d:
    print(part)
```

**Explanation:**
- `np.hsplit()`  divides the columns into equal sections.

## <span style="color: skyblue;"> Vertical Splitting with np.vsplit() </span>

`np.vsplit()` splits a 2D array vertically into sections (rows).

>**Shortcut:** Equivalent to np.split(array, N, axis=0).

**Example: Vertical Splitting**

```python
# Vertical split
vsplit_2d = np.vsplit(array_2d, 2)

print("Original Array:\n", array_2d)
print("Vertical Splits:")
for part in vsplit_2d:
    print(part)
```

**Explanation:**
* `np.vsplit()` divides rows into equal sections.
* This is a convenient method for row-wise splitting.

## <span style="color: skyblue;"> Custom Splits with Indices </span>

Instead of dividing into equal parts, we can use indices for custom splits.

```python
# Split using indices
custom_split = np.split(array_1d, [2, 5])

print("Original Array:", array_1d)
print("Custom Splits:", custom_split)
```

**Explanation:**
- Splits occur at positions `[2, 5]`, creating 3 parts:
  - `[1, 2]`
  - `[3, 4, 5]`
  - `[6]`.

<span style="color: red;"> Note:</span>
* This is based on index positions, not actual values
* It is less commonly used, but learners should be aware of it

## <span style="color: skyblue;"> Stacking in NumPy </span>

Stacking is the opposite of splitting — instead of dividing arrays, we combine them.

## <span style="color: skyblue;"> Vertical Stacking with np.vstack() </span>

`np.vstack()` stacks arrays along the vertical axis (rows).
* All arrays must have the same number of columns for vertical stacking.

> **[Interactive]:** Watch vstack and hstack combine arrays visually:  
> [Stacking Visualizer](https://aiml-v-izs.vercel.app/v/fd-z1y6t8h5/stacking?embed=1)

**Example: Vertical Stacking**

```python
import numpy as np

# Define arrays
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

# Stack vertically
vstack_result = np.vstack([array1, array2])

print("Array 1:", array1)
print("Array 2:", array2)
print("Vertical Stack:\n", vstack_result)
```

**Explanation:**
- The arrays `[1, 2, 3]` and `[4, 5, 6]` are stacked row-wise to form a 2D array.

`np.vstack()` requires arrays to have the same number of columns.

Let's see what happens when this condition is not met.

```python
# Arrays with unequal sizes
# This is expected to fail!
array3 = np.array([7, 8])

vstack_error = np.vstack([array1, array3])
```

**Explanation:**
- Vertical stacking fails because `array3` has fewer elements than `array1`.
- All arrays must have the same number of columns for `np.vstack()`.



## <span style="color: skyblue;"> Horizontal Stacking with np.hstack() </span>

`np.hstack()` stacks arrays along the horizontal axis (columns).
* All arrays must have the same number of rows for horizontal stacking.

> **[Interactive]:** Watch vstack and hstack combine arrays visually:  
> [Stacking Visualizer](https://aiml-v-izs.vercel.app/v/fd-z1y6t8h5/stacking?embed=1)\
> Select horizontal from dropdown

**Example: Horizontal Stacking**

```python
# Define arrays
array4 = np.array([[1], [2], [3]])
array5 = np.array([[4], [5], [6]])

# Stack horizontally
hstack_result = np.hstack([array4, array5])

print("Array 4:\n", array4)
print("Array 5:\n", array5)
print("Horizontal Stack:\n", hstack_result)
```

**Explanation:**
- Arrays `array4` and `array5` are stacked column-wise to form a 2D array.

`np.hstack()` requires arrays to have the same number of rows.

Let's see what happens when this condition is not met.

```python
# Arrays with unequal rows
# This is expected to fail!
array6 = np.array([[7], [8]])

hstack_error = np.hstack([array4, array6])
```

**Explanation:**
- Horizontal stacking fails because `array6` has fewer rows than `array4`.
- All arrays must have the same number of rows for `np.hstack()`.

### Real world use cases

#### Vertical stacking

Zomato already has a matrix of restaurants — when new restaurants join later, instead of rebuilding the whole matrix, add new rows using `vstack`.

#### Horizontal stacking

Adding a new feature column — for instance, a popularity score — can be attached to the existing restaurant dataset using `hstack`.

<span style="color: red;"> Important:</span>
* Shapes must be compatible
* Number of columns should align for vertical stacking
* Number of rows should align for horizontal stacking

### Business Context

In Zomato analytics, we often work with multiple datasets or features that need to be combined for comprehensive analysis. Stacking arrays allows us to merge data along rows or columns efficiently:

- Vertical stacking (np.vstack): Combine multiple datasets of the same features (columns) but different restaurants (rows). Example: Adding new restaurant entries to an existing dataset.
- Horizontal stacking (np.hstack): Merge different features (columns) for the same set of restaurants (rows). Example: Combining votes, ratings, and approximate costs into a single 2D array for analysis.
- Ensures data alignment and consistency, which is crucial for downstream tasks like scoring, filtering, or visualization.
- Helps maintain structured, tabular data while performing feature engineering, aggregations, or building predictive models.


## <span style="color: skyblue;"> General Concatenation with np.concatenate() </span>

`np.concatenate()` concatenates arrays along any axis (default is `axis=0`).  
It provides more flexibility than stacking functions.

**Parameters:**
1. **array:** List or tuple of arrays to concatenate.
2. **axis:** Axis along which to concatenate.
   - `axis=0`: Concatenate rows.
   - `axis=1`: Concatenate columns (for 2D arrays).

```python
# Example 1: Concatenation Along Rows
# Concatenate along axis=0 (rows)
concat_rows = np.concatenate([array4, array5], axis=0)

print("Concatenation Along Rows:\n", concat_rows)
```

**Explanation:**
- Arrays `array4` and `array5` are concatenated row-wise into a single array.

```python
# Example 2: Concatenation Along Columns
# Concatenate along axis=1 (columns)
concat_columns = np.concatenate([array4, array5], axis=1)

print("Concatenation Along Columns:\n", concat_columns)
```

**Explanation:**
- Arrays `array4` and `array5` are concatenated column-wise into a single array.

`np.concatenate()` requires compatible dimensions along the non-concatenating axis.

Let's see what happens when this condition is not met.

```python
# Arrays with incompatible shapes
# This is expected to fail!
array7 = np.array([7, 8, 9])

concat_error = np.concatenate([array4, array7], axis=1)
```

**Explanation:**
- Concatenation along `axis=1` fails because `array7` does not match the row count of `array4`.
- Always ensure compatible dimensions for `np.concatenate()`.

<span style="color: red;"> Note:</span>
* `concatenate` needs explicit axis handling and is usually less intuitive for beginners
* In most learning cases, `vstack` and `hstack` are easier to understand and use

### Business Context

In restaurant analytics, np.concatenate() enables efficient integration of datasets across restaurants, features, or time periods. By combining rows or columns, it supports comprehensive analyses, feature engineering, and predictive modeling, ensuring aligned, complete datasets for accurate business insights.

# Introduction to Pandas

Pandas is not that cute animal panda.

Pandas is a powerful Python library used for working with structured data. It builds on top of **NumPy** and provides easy-to-use data structures and functions for handling structured data.

The easiest way to understand Pandas is:

* NumPy gives us arrays
* Pandas builds on top of that
* Pandas adds structure and metadata

So the conceptual flow is:

* NumPy array → Pandas Series → Pandas DataFrame

### What is missing in NumPy?

Suppose we create a NumPy array.

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])
arr
```

This gives us values. But what is missing?

* There is no column name
* There is no meaningful explicit row label
* It is not naturally table-like
* It does not directly feel like an Excel sheet

Yes, NumPy elements do have positions such as `0`, `1`, `2`, but those are implicit positions. They are not rich labels that make the data more meaningful.

### From NumPy to Series

When we convert this idea into a Pandas Series, we still have values, but now we also get metadata — a column name, explicit indexes, and better structure for reading and analysis.

A Series is:
* Still one dimensional
* Still efficient
* Still compatible with many NumPy style ideas
* But much more meaningful because it has labels

### From Series to DataFrame

If one labeled column is a Series, then multiple labeled columns together become a DataFrame. A DataFrame is table shaped — it feels like an Excel sheet, a database table, or a CSV loaded into memory. This is why Pandas becomes extremely useful in data work.

### Why Pandas is used in practice

In the real world, we do not usually receive neat vectors one by one. We usually receive data in files such as CSV, Excel, extracted reports, product logs, or tables from a database. And once data arrives, it often has problems such as missing values, mixed data types, corrupted entries, and inconsistent formatting.

So we need a library that helps us:
* Read the data
* Inspect the data
* Clean the data
* Transform the data
* Analyze the data
* Visualize the data

That is why Pandas is so important.

### Why Use Pandas?
1. **Real-World Relevance**: Pandas helps analyze large datasets efficiently, such as customer reviews, financial data, or restaurant data like our **Zomato Bangalore dataset**.
2. **Core Strengths**:
   - Data cleaning, preparation, and exploration.
   - Supports handling **missing data**.
   - Integrates seamlessly with libraries like `Matplotlib` and `NumPy`.

## Pandas Core Components - DataFrame and Series

Pandas has two core structures: `Series` and `DataFrame`. These two ideas must be extremely clear — everything else in Pandas builds on top of them.

### Series

A **Series** is a one-dimensional labeled array that can hold any data type. Think of it as a **single column of data** in a table.

A Pandas `Series` contains:
* Values
* An index
* Optionally a name

This makes it more powerful than a plain NumPy array.

**Mental model:** Think of it like this:
* NumPy array = values
* Series = values + labels + metadata

That extra metadata is the big idea. It allows us to work with data more naturally.

A Series can be imagined like a single Excel column: one column name, several row labels, and values aligned against those labels. That is why it feels much closer to real data than a plain vector.

**Key Features for Zomato Analysis**:
- Data is labeled using an **index** (e.g., restaurant names or IDs for easy lookup).
- Supports operations like indexing, slicing, and filtering (e.g., selecting restaurants with ratings above 4.0 or costs below Rs 500).

### DataFrame

A **DataFrame** is a two-dimensional table with labeled rows and columns. It is the core structure in Pandas, ideal for tabular data like the Zomato dataset.

If one labeled column is a Series, then multiple labeled columns together become a DataFrame. Each column can be thought of as a Series — so if multiple Series come together in aligned tabular form, we get a DataFrame.

**Key Features for Zomato Analysis**:
- Can hold data of different types (e.g., restaurant names as strings, ratings as floats, cost_for_two as integers).
- Labeled rows (**index**, e.g., restaurant IDs) and columns (e.g., cuisine, location).
- Provides tools for filtering (e.g., restaurants in a specific area), aggregating (e.g., average rating by cuisine), and transforming data (e.g., normalizing costs).

`iloc` stands for integer location based indexing. It uses row positions — meaning it works with the implicit row numbers like `0`, `1`, `2`, `3`. Even if those row numbers are not being emphasized visually, they still exist internally.

## Adding a New Column

Add a new column by assigning data to a new column name.

### Problem with the ratings column

The ratings are not stored as plain numbers. They appear in forms such as `4.1/5` or `3.9/5`. This is a problem — if we want to calculate average rating, maximum rating, or filter on rating, the values must be numeric. Right now they are strings.

We want to remove the `/5` part and keep only the numeric value. We do this using string slicing — starting from index `0` and stopping before the last two characters. So `4.1/5` becomes `4.1`.

```python
# Adding a new column
df["only_rating"] = df["Ratings"].str[:-2]

# Displaying a new column
df["only_rating"]
```

This creates a new column called `only_rating`. When you create a new column like this, Pandas adds it as a new column in the DataFrame. By default, newly created columns are added at the end — so if the DataFrame originally had 17 columns, this becomes the 18th column.

As you can see, the rating is still in object type. We need to convert it into a float type.

```python
# It is expected to fail!
df["only_rating"] = df["only_rating"].astype(float)
```

<span style="color: violet;">It is failing because it looks like there is an 'N' in the ratings column, which cannot be converted into a float. What are our options?</span>

This is a common real world data cleaning scenario — always assume that some amount of real world data is corrupted or inconsistent. Some rows may contain values like `NEW/5`. After slicing, this may still result in invalid strings such as `NEW`. If we try to convert that directly to float, it fails.

### Option 1

**Replacing certain values in object type columns**

```python
df["only_rating"] = df["only_rating"].replace('N', np.nan)
```

```python
df["only_rating"] = df["only_rating"].astype(float)
```

It is now working

### Option 2

**Use pd.to_numeric with errors='coerce'**

`pd.to_numeric` converts a column into a numeric type. `errors='coerce'` is a parameter that will force the operation and convert values it cannot handle into nulls. It tells Pandas: if conversion is valid, convert it; if conversion fails, turn it into a missing value. Instead of crashing, Pandas handles them gracefully.

```python
df["only_rating"] = pd.to_numeric(df["only_rating"], errors='coerce')
```

```python
df["only_rating"]
```

```python
df["cost_for_two"]
```

```python
df["cost_for_two"] = pd.to_numeric(df["cost_for_two"].str.strip("'\""), errors="coerce")
```

```python
df["cost_for_two"]
```

If `cost_for_two` remains text, then operations like sorting by cost, average cost, dividing by 2, or filtering expensive restaurants will not work properly.

```python
# Adding a new column
df['cost_for_one']  = df['cost_for_two'] / 2

# Displaying a new column
df['cost_for_one']
```
